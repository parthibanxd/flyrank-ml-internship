## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 content items across 32 clients. Of these, 16,262 items (54.2%) have the starter declining label, which gives us a substantial set of pages that may warrant investigation. The median 90-day search impressions is 731, so the dataset contains measurable search visibility rather than only very low-traffic pages. These numbers make refresh prioritization worth investigating, although they do not show that a refresh will cause future improvement.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/parthibanxd/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print(f"Rows: {len(df):,}")
print(f"Clients: {df['client_id'].nunique():,}")

declining = (df["trend_direction"] == "down").sum()
print(f"Declining items: {declining:,}")

visible = df[df["impressions_90d"] > 0]

print(f"Items with search impressions: {len(visible):,}")
print(
    f"Median impressions among visible items: "
    f"{visible['impressions_90d'].median():.1f}"
)
```
